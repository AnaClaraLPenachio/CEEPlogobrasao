```python
# tipos de dados no python - revisão

#int
idade = 17
print("idade: ")
print(idade)
print(type(idade))

#float valores com virgula
altura = 1.65
print("altura:")
print(altura)
print(type(altura))

#str (string) texto
nome = "Charlie"
print("nome:")
print(nome)
print(type(nome))

#bool - booleano (sim, não, verdade, mentira, ligado, desligado) apenas opções de correto ou incorreto
briga = False
print("briga: ")
print(briga)
print(type(briga))

#list Lista (coleção ordenada)
frutas = ["blueberry", "morango", "manga",]
print("frutas: ")
print(frutas)
print(type(frutas))

#tuple - dupla, como por exemplo, coordenadas
coordenadas = (52.640, 24.325)
print("coordenadas: ")
print(coordenadas)
print(type(coordenadas))

#set - conjunto
numeros = {1,2,3,4,5,6,7,8,9,}
print("numeros: ")
print(numeros)
print(type(numeros))

#dict - dicionario
pessoas = {"nome": "Charles", "idade": 17}
print("pessoas: ")
print(pessoas)
print(type(pessoas))

#None Type = quando não tem nada, mas preciso informar um valor
valor = None
```
